In clustering, there’s no direct “accuracy” measure like in supervised learning, since we don’t have true labels to compare. However, there are several ways to evaluate the quality of clusters in K-Means clustering:

1. Silhouette Score

	•	The silhouette score measures how similar data points are within a cluster compared to other clusters. Scores range from -1 to 1, where values closer to 1 indicate well-separated clusters.
	•	A higher silhouette score generally indicates better-defined clusters.

2. Inertia (Within-Cluster Sum of Squares)

	•	Inertia is the sum of squared distances from each point to its assigned cluster center. Lower inertia indicates tighter clusters, which is usually desirable.
	•	You’ve already calculated inertia in the elbow method, which helps determine the optimal number of clusters.
	•	This metric is not bounded and generally decreases as the number of clusters increases, so it’s useful primarily for comparing between models with different numbers of clusters.

3. Calinski-Harabasz Index (Variance Ratio Criterion)

	•	The Calinski-Harabasz Index measures the ratio of the sum of between-cluster dispersion to within-cluster dispersion. A higher score suggests better-defined clusters.

4. Davies-Bouldin Index

	•	The Davies-Bouldin Index calculates the average similarity ratio of each cluster with its most similar cluster, where a lower score is better (indicating more compact clusters).

# Step 6

You will see frequently that loops are used to test different algorithms. We will do that next.

Let's review other clustering methods:

- K-Means: This method tries to divide points into a set number of groups (clusters) by placing “centers” and assigning each point to the nearest center. K-Means works well with evenly sized, round clusters but struggles with groups of different shapes or sizes and with noisy data.

- DBSCAN: This method finds groups (clusters) by looking for areas where points are close together. It’s great for finding oddly shaped groups and can ignore “noise” or outlier points that don’t fit into any cluster.

- Agglomerative Clustering: This method starts by treating each point as its own cluster, then slowly merges the closest clusters together step-by-step until there are just a few clusters left. It’s flexible and doesn’t assume clusters are any specific shape.

In short, DBSCAN is best for finding oddly shaped clusters, Agglomerative for gradually building clusters, and K-Means for simple, round clusters.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import pandas as pd

# Standardize the data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_numeric)

# Define clustering algorithms to test
clustering_algorithms = {
    "KMeans": KMeans(n_clusters=3, random_state=42),
    "AgglomerativeClustering": AgglomerativeClustering(n_clusters=3),
    "DBSCAN": DBSCAN()
}

# Function to evaluate clustering algorithms
def evaluate_clustering(data, models):
    results = []
    
    for name, model in models.items():
        # Fit the model and obtain cluster labels using fit_predict()
        cluster_labels = model.fit_predict(data)
        
        # Calculate metrics
        silhouette_avg = silhouette_score(data, cluster_labels)
        ch_score = calinski_harabasz_score(data, cluster_labels)
        db_score = davies_bouldin_score(data, cluster_labels)
        
        # Append results
        results.append({
            "Algorithm": name,
            "Silhouette Score": silhouette_avg,
            "Calinski-Harabasz Score": ch_score,
            "Davies-Bouldin Score": db_score
        })
        
    # Convert results to DataFrame for better readability
    return pd.DataFrame(results)

# Run the evaluation
evaluation_results = evaluate_clustering(data_scaled, clustering_algorithms)

# Display the results
print(evaluation_results)
```
